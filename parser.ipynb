```python
from bs4 import BeautifulSoup
import pandas as pd
from collections import defaultdict
import re
```

```python
with open("html.html") as fp:
    soup = BeautifulSoup(fp, 'html.parser')
```

```python
articles = soup.find_all('article')

result = defaultdict(list)
for article in articles:
	result['name'].append(article.find('p', class_='item-name').text.replace('\n', '').replace('(', '').replace(')','').strip().split(' ')[0])
	result['age'].append(re.search('\d+', article.find('p', class_='item-name').text).group())
	result['item_title'].append(article.find('h3', class_='item-title').text)
	result['item_text'].append(article.find('div', class_='item-text').find('p').text.replace('\n', ' ').strip())
```

```python
pd.DataFrame.from_dict(result).to_csv('result.csv')
```
